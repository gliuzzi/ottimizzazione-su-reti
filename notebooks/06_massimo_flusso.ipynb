## Obiettivi

Calcolare un flusso massimo, verificare la conservazione e mostrare il taglio minimo associato.

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G=nx.DiGraph()
for i,j,u in [('s','a',7),('s','b',4),('a','b',3),('a','t',5),('b','t',6)]:
    G.add_edge(i,j,capacity=u)
value, flow = nx.maximum_flow(G,'s','t',capacity='capacity')
cut_value, (S,T) = nx.minimum_cut(G,'s','t',capacity='capacity')
print('Flusso massimo:',value)
print('Taglio minimo:',cut_value,' S =',S,' T =',T)
assert value == cut_value
```

```python
for v in set(G)-{'s','t'}:
    incoming=sum(flow[u][v] for u in G.predecessors(v))
    outgoing=sum(flow[v][w] for w in G.successors(v))
    print(v, incoming, outgoing)
    assert incoming == outgoing
```

```python
pos={'s':(0,0),'a':(1,1),'b':(1,-1),'t':(2,0)}
labels={(i,j):f'{flow[i][j]}/{G[i][j]["capacity"]}' for i,j in G.edges}
colors=['#d8ecff' if v in S else '#ffe0e0' for v in G]
nx.draw_networkx(G,pos,node_color=colors,arrows=True)
nx.draw_networkx_edge_labels(G,pos,edge_labels=labels)
plt.title('Flusso/capacità; colori = lati del taglio'); plt.axis('off')
```
