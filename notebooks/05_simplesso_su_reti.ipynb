## Obiettivi

Richiamare la struttura di una base ad albero e usare `network_simplex`, osservando soluzione, archi positivi e vincoli attivi. NetworkX restituisce la soluzione finale, non le singole iterazioni pivot.

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.DiGraph()
for nodo, domanda in {'1':-6, '2':0, '3':2, '4':4}.items():
    G.add_node(nodo, demand=domanda)
archi=[('1','2',5,2),('1','3',4,4),('2','3',3,1),('2','4',4,3),('3','4',5,1)]
for i,j,u,c in archi:
    G.add_edge(i,j,capacity=u,weight=c)
```

```python
costo, flow = nx.network_simplex(G, demand='demand', capacity='capacity', weight='weight')
print('Costo ottimo:', costo)
positivi=[]
for i,j in G.edges:
    x=flow[i][j]
    stato='inferiore' if x==0 else ('superiore' if x==G[i][j]['capacity'] else 'interno')
    if x>0: positivi.append((i,j))
    print(f'({i},{j}): x={x}, stato={stato}')
print('Archi con flusso positivo:', positivi)
```

## Base ad albero

In una soluzione di base non degenere, gli archi di base formano un albero ricoprente del grafo sottostante. Gli archi a un estremo possono essere non di base. Dal solo flusso finale, soprattutto in presenza di degenerazione, la base scelta internamente dall'algoritmo non è identificabile in modo univoco.

```python
supporto = nx.Graph(); supporto.add_nodes_from(G.nodes); supporto.add_edges_from(positivi)
print('Il supporto positivo è connesso?', nx.is_connected(supporto))
print('Contiene cicli?', len(nx.cycle_basis(supporto)) > 0)
```
