## Obiettivi

Riconoscere un albero, calcolare un albero ricoprente minimo e ordinare topologicamente un DAG.

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.Graph()
G.add_weighted_edges_from([('a','b',4), ('a','c',2), ('b','c',1), ('b','d',5), ('c','d',3), ('c','e',6), ('d','e',2)])
T = nx.minimum_spanning_tree(G, weight='weight')
print('È un albero?', nx.is_tree(T))
print('Archi:', sorted(T.edges(data='weight')))
print('Peso totale:', T.size(weight='weight'))
```

```python
pos = nx.spring_layout(G, seed=4)
plt.figure(figsize=(7,4))
nx.draw_networkx(G, pos, node_color='#eaf2ff', edge_color='#bbbbbb')
nx.draw_networkx_edges(G, pos, edgelist=T.edges(), width=3, edge_color='#c62828')
nx.draw_networkx_edge_labels(G, pos, edge_labels=nx.get_edge_attributes(G, 'weight'))
plt.title('Albero ricoprente minimo in rosso'); plt.axis('off')
```

## Grafi orientati aciclici

Un ordinamento topologico dispone ogni nodo dopo tutti i suoi predecessori.

```python
D = nx.DiGraph([('A','C'), ('B','C'), ('B','D'), ('C','E'), ('D','E')])
print('È aciclico?', nx.is_directed_acyclic_graph(D))
print('Un ordinamento topologico:', list(nx.topological_sort(D)))
print('Generazioni:', list(nx.topological_generations(D)))
```
