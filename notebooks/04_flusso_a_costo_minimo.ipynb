## Convenzione delle domande

Adottiamo la convenzione NetworkX: $d_i>0$ per un nodo di domanda e $d_i<0$ per un nodo di offerta. Il bilancio è
$$\sum_{j:(j,i)\in A}x_{ji}-\sum_{j:(i,j)\in A}x_{ij}=d_i.$$

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.DiGraph()
G.add_node('s', demand=-5)
G.add_node('a', demand=0)
G.add_node('b', demand=0)
G.add_node('t', demand=5)
G.add_edge('s','a', capacity=4, weight=2)
G.add_edge('s','b', capacity=3, weight=1)
G.add_edge('a','b', capacity=2, weight=1)
G.add_edge('a','t', capacity=4, weight=1)
G.add_edge('b','t', capacity=4, weight=3)
assert sum(nx.get_node_attributes(G, 'demand').values()) == 0
```

```python
flow = nx.min_cost_flow(G, demand='demand', capacity='capacity', weight='weight')
cost = nx.cost_of_flow(G, flow, weight='weight')
print('Costo ottimo:', cost)
for i, j in G.edges:
    print(f'x[{i},{j}] = {flow[i][j]} / {G[i][j]["capacity"]}')
```

```python
for i in G.nodes:
    entrante = sum(flow[j][i] for j in G.predecessors(i))
    uscente = sum(flow[i][j] for j in G.successors(i))
    print(i, 'entrante - uscente =', entrante-uscente, 'domanda =', G.nodes[i]['demand'])
```

```python
pos={'s':(0,0),'a':(1,1),'b':(1,-1),'t':(2,0)}
labels={(i,j):f'{flow[i][j]}/{G[i][j]["capacity"]}; c={G[i][j]["weight"]}' for i,j in G.edges}
nx.draw_networkx(G,pos,node_color='#eaf2ff',arrows=True)
nx.draw_networkx_edge_labels(G,pos,edge_labels=labels,font_size=9)
plt.title(f'Flusso ottimo — costo {cost}'); plt.axis('off')
```
