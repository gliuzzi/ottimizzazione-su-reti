# Primi passi con NetworkX
Costruiamo una piccola rete e calcoliamo un cammino minimo.

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.DiGraph()
G.add_weighted_edges_from([("s", "a", 4), ("s", "b", 2), ("b", "a", 1), ("a", "t", 3), ("b", "t", 7)])
cammino = nx.shortest_path(G, "s", "t", weight="weight")
lunghezza = nx.shortest_path_length(G, "s", "t", weight="weight")
cammino, lunghezza
```

```python
pos = nx.spring_layout(G, seed=7)
nx.draw_networkx(G, pos, node_color="#eef4ff", edge_color="#555555")
nx.draw_networkx_edge_labels(G, pos, edge_labels=nx.get_edge_attributes(G, "weight"))
plt.axis("off");
```
