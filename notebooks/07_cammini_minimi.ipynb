## Dijkstra

Con pesi non negativi possiamo applicare l'algoritmo di Dijkstra.

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G=nx.DiGraph()
G.add_weighted_edges_from([('s','a',4),('s','b',2),('b','a',1),('a','c',3),('b','c',5),('c','t',2),('a','t',7)])
dist, paths = nx.single_source_dijkstra(G,'s',weight='weight')
for v in dist:
    print(v, 'distanza =', dist[v], 'cammino =', paths[v])
```

```python
path=paths['t']; path_edges=list(zip(path,path[1:]))
pos=nx.spring_layout(G,seed=10)
nx.draw_networkx(G,pos,node_color='#eaf2ff',arrows=True)
nx.draw_networkx_edges(G,pos,edgelist=path_edges,width=3,edge_color='#c62828')
nx.draw_networkx_edge_labels(G,pos,edge_labels=nx.get_edge_attributes(G,'weight'))
plt.title(f'Cammino minimo s-t, lunghezza {dist["t"]}'); plt.axis('off')
```

## Pesi negativi

Dijkstra non è adatto in presenza di archi di peso negativo. Se non esistono cicli negativi si può usare Bellman–Ford.

```python
H=nx.DiGraph()
H.add_weighted_edges_from([('s','a',3),('s','b',5),('a','b',-2),('a','t',6),('b','t',2)])
print(nx.single_source_bellman_ford(H,'s',weight='weight'))
print('Ciclo negativo?', nx.negative_edge_cycle(H,weight='weight'))
```
