## Obiettivi

Costruire grafi orientati e non orientati, ispezionare nodi e archi e distinguere un sottografo generico da un sottografo indotto.

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.Graph()
G.add_edges_from([(1, 2), (1, 3), (2, 3), (2, 4), (3, 4)])
print('Nodi:', list(G.nodes))
print('Spigoli:', list(G.edges))
print('Gradi:', dict(G.degree()))
```

## Sottografo e sottografo indotto

Fissato $W=\{1,2,3\}$, un sottografo può scegliere solo alcuni spigoli con estremi in $W$. Il sottografo indotto $G[W]$ deve conservarli tutti.

```python
W = {1, 2, 3}
H = nx.Graph([(1, 2), (2, 3)])          # sottografo non indotto
GW = G.subgraph(W).copy()               # sottografo indotto
print('Archi di H:', sorted(H.edges()))
print('Archi di G[W]:', sorted(GW.edges()))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 3.5))
posG = {1:(0,0), 2:(1,1), 3:(1,-1), 4:(2,0)}
nx.draw_networkx(G, posG, ax=axes[0], node_color='#eaf2ff'); axes[0].set_title('Grafo G')
posW = {k: posG[k] for k in W}
nx.draw_networkx(H, posW, ax=axes[1], node_color='#fff0f0'); axes[1].set_title('Sottografo H')
nx.draw_networkx(GW, posW, ax=axes[2], node_color='#eaffea'); axes[2].set_title('Indotto G[W]')
for ax in axes: ax.axis('off')
plt.tight_layout()
```
