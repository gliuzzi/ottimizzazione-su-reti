## 1. Riconoscimento della struttura

Per il grafo seguente: determinare componenti connesse, eventuali cicli e un albero ricoprente minimo. Completare le celle e poi confrontare con la soluzione proposta.

```python
import networkx as nx
import matplotlib.pyplot as plt
G=nx.Graph()
G.add_weighted_edges_from([(1,2,3),(1,3,2),(2,3,1),(2,4,5),(3,4,4),(4,5,2)])
```

```python
# Soluzione
print('Componenti:', list(nx.connected_components(G)))
print('Cicli:', nx.cycle_basis(G))
T=nx.minimum_spanning_tree(G)
print('MST:', list(T.edges(data='weight')), 'peso =', T.size(weight='weight'))
```

## 2. Modellazione di un trasporto

Un deposito offre 8 unità; due clienti ne richiedono rispettivamente 3 e 5. Costruire la rete, calcolare il flusso di costo minimo e verificare i bilanci.

```python
R=nx.DiGraph()
for n,d in {'D':-8,'H':0,'C1':3,'C2':5}.items(): R.add_node(n,demand=d)
for i,j,u,c in [('D','H',8,1),('D','C1',3,4),('H','C1',4,1),('H','C2',6,2)]:
    R.add_edge(i,j,capacity=u,weight=c)
flow=nx.min_cost_flow(R)
print('Costo:',nx.cost_of_flow(R,flow),'Flusso:',flow)
```

## 3. Massimo flusso e taglio

Calcolare massimo flusso e taglio minimo della rete seguente. Modificare poi una sola capacità per aumentare il valore ottimo.

```python
F=nx.DiGraph()
for i,j,u in [('s','a',5),('s','b',6),('a','b',2),('a','t',4),('b','t',5)]: F.add_edge(i,j,capacity=u)
print('Massimo flusso:',nx.maximum_flow_value(F,'s','t'))
print('Taglio minimo:',nx.minimum_cut(F,'s','t'))
```

## 4. Cammini minimi

Calcolare le distanze da `s`, ricostruire un cammino minimo fino a `t` e spiegare quale algoritmo è appropriato.

```python
P=nx.DiGraph()
P.add_weighted_edges_from([('s','a',2),('s','b',6),('a','b',1),('a','t',5),('b','t',2)])
print(nx.single_source_dijkstra(P,'s'))
```
